# Create symmetric edges
w_bin_edges = np.linspace(w_0 - half_range, w_0 + half_range, num_bins + 1)


# --- Step 3: For each bin, find indices in ws that fall in the bin ---
pos_unbinned = np.copy(pos_lin)
neg_unbinned = np.copy(neg_lin)

for b in range(num_bins):
    bin_start = w_bin_edges[b]
    bin_end = w_bin_edges[b+1]

    # Get indices within this bin
    bin_indices = np.where((ws >= bin_start) & (ws < bin_end))[0]
    if len(bin_indices) == 0:
        continue  # skip empty bins
    
    # Average phase within this bin
    pos_avg = np.mean(pos_lin[bin_indices])
    neg_avg = np.mean(neg_lin[bin_indices])



    # Assign binned phase
    pos_lin[bin_indices] = pos_avg
    neg_lin[bin_indices] = neg_avg

neg_lin = -pos_lin
#pos_lin = pos_binned
#neg_lin = neg_binned

SLM_mask = (ws >= w_0 - half_range) & (ws <= w_0 + half_range)
pos_lin[~SLM_mask] = 0
neg_lin[~SLM_mask] = 0

# --- Compute edge values of the binned region ---
left_edge_val_pos = pos_lin[SLM_mask][0]
right_edge_val_pos = pos_lin[SLM_mask][-1]

left_edge_val_neg = neg_lin[SLM_mask][0]
right_edge_val_neg = neg_lin[SLM_mask][-1]

# --- Apply edge extension instead of zeroing ---
pos_lin[ws < w_0 - half_range] = left_edge_val_pos
pos_lin[ws > w_0 + half_range] = right_edge_val_pos

neg_lin[ws < w_0 - half_range] = left_edge_val_neg
neg_lin[ws > w_0 + half_range] = right_edge_val_neg
